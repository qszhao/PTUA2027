```python
import random
import math
import statistics
```

```python
def simulate_pi(criterion):
    n = 0
    d = 0

    while True:
        x = random.random()
        y = random.random()

        if x**2 + y**2 <= 1.0:
            n += 1

        d += 1
        estimated_pi = 4 * n / d
        relative_error = abs(estimated_pi - math.pi) / math.pi

        if relative_error <= criterion:
            return estimated_pi, d
```

```python
criteria = [0.01, 0.001, 0.0001, 0.00001]
results = {}

for criterion in criteria:
    draw_counts = []
    estimates = []

    print(f"\nConvergence criterion: {criterion}", flush=True)

    for run in range(1, 11):
        estimated_pi, draws = simulate_pi(criterion)

        estimates.append(estimated_pi)
        draw_counts.append(draws)

        print(
            f"Run {run}: pi = {estimated_pi:.10f}, draws = {draws}",
            flush=True
        )

    results[criterion] = {
        "estimates": estimates,
        "draw_counts": draw_counts
    }

    print(f"Average draws: {statistics.mean(draw_counts):.2f}")
    print(f"Standard deviation: {statistics.stdev(draw_counts):.2f}")
```

## Experiment notes

For each relative error tolerance (0.01, 0.001, 0.0001, and 0.00001),
10 independent simulations were performed.

Each simulation stopped when:
abs(estimated_pi - math.pi) / math.pi <= criterion

No maximum draw limit was imposed.

The mean and sample standard deviation of the draw counts were
calculated for each tolerance. The sample standard deviation uses
a denominator of n - 1.

Because the simulation is random, results vary between runs.
