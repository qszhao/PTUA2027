```python
import math
import random
import statistics

pi = math.pi

criteria = [0.01, 0.001, 0.0001, 0.00001]
runs = 10

for crit in criteria:
    draws_record = []

    for r in range(runs):
        n = 0
        d = 0
        while True:
            x = random.random()
            y = random.random()
            if x**2 + y**2 <= 1.0:
                n += 1
            d += 1

            ratio = 4.0 * n / d
            pct_diff = abs(ratio - pi) / pi * 100

            if pct_diff <= crit:
              draws_record.append(d)
              break
avg_draws = statistics.mean(draws_record)
std_draws = statistics.stdev(draws_record)

print(f"Convergence criterion: {crit}%")
print(f"  ten draws: {draws_record}")
print(f"  Average draws: {avg_draws:,.2f}")
print(f"  Std deviation: {std_draws:,.2f}")
print("-" * 50)
```
