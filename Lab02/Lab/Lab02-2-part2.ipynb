```python
import math
import random
import statistics


def draws_to_tolerance(tolerance):
    inside = 0
    draws = 0
    relative_error = math.inf

    while relative_error > tolerance:
        x, y = random.random(), random.random()
        if x*x + y*y <= 1:
            inside += 1
        draws += 1
        estimate = 4 * inside / draws
        relative_error = abs(estimate - math.pi) / math.pi

    return draws
```

```python
print(draws_to_tolerance(0.01))
```

```python
tolerances = [0.01, 0.001, 0.0001, 0.00001]

for tolerance in tolerances:
    counts = [draws_to_tolerance(tolerance) for _ in range(10)]
    mean = statistics.mean(counts)
    sd = statistics.stdev(counts)
    print(f'Tolerance {tolerance:g}: mean = {mean:.1f}, sample SD = {sd:.1f} draws', flush=True)
```
