```python
def is_anagram(word1,word2):
    return sorted (word1)==sorted(word2)
print(is_anagram("takes", "steak"))
```

```python
from urllib.request import urlopen

url = 'https://raw.githubusercontent.com/AllenDowney/ThinkPython/v3/words.txt'
with urlopen(url) as response:
    word_list = response.read().decode('utf-8').split()

print(len(word_list))
```

```python
for word in word_list:
    if word != "takes" and is_anagram(word, "takes"):
        print(word)
```

```python
def reverse_sentence(sentence):
    words = sentence.split()
    words.reverse()
    return " ".join(words).capitalize()

print(reverse_sentence("Reverse this sentence"))
```

```python
def has_duplicates(sequence):
    seen = {}
    for item in sequence:
        if item in seen:
            return True
        seen[item] = True
    return False

print(has_duplicates("banana"))
print(has_duplicates("skate"))
```

```python
unique_words = [word for word in word_list if not has_duplicates(word)]
longest_length = max(len(word) for word in unique_words)

for word in unique_words:
    if len(word) == longest_length:
        print(word, len(word))
```
