```python
#Exercise 9.15.2

def is_anagram(x,y):
    xSorted = sorted(x)
    ySorted = sorted(y)

    if xSorted == ySorted:
        return True
    else:
        return False

print(is_anagram("tops","stop"))
print(is_anagram("tops","hello"))
```

```python
#Exercise 10.11.3
def has_duplicates(word):
    seen = {}

    for element in word:
        if element in seen:
            return True

        seen[element] = 1

    return False

has_duplicates("banana")
has_duplicates("unpredictably")
has_duplicates([1, 2, 3, 4])
```
