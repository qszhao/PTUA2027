# Lab02-2 Task 2: Pi Simulation

Run 10 simulations for each convergence criterion (`0.01`, `0.001`, `0.0001`, `0.00001`). Print the average and standard deviation of the number of draws. No draw limit.

```python
import random
import math
import statistics
```

```python
pi = math.pi
criteria = [0.01, 0.001, 0.0001, 0.00001]
results = {}

for criterion in criteria:
    draws = []
    estimates = []

    for run in range(10):
        n = 0 # number of points falling in the unit circle
        d = 0 # number of points falling in the unit square

        while True:
            x = random.random()
            y = random.random()
            if x**2 + y**2 <= 1.0:
                n += 1
            d += 1
            ratio = 4 * n / d

            if abs(ratio - pi) / pi <= criterion:
                draws.append(d)
                estimates.append(ratio)
                break

    results[criterion] = {"draws": draws, "estimates": estimates}

print(f"{'Criterion':<12} {'Average draws':>15} {'Std. deviation':>18}")
for criterion in criteria:
    draws = results[criterion]["draws"]
    print(f"{criterion:<12g} {statistics.mean(draws):>15.2f} {statistics.stdev(draws):>18.2f}")
```
