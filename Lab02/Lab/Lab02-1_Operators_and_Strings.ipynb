# Lists

## 9.15.2. Exercise

- Write an `is_anagram` function. Use the word list to find the anagrams of `takes`.

```python
def is_anagram(word1, word2):
    return sorted(word1) == sorted(word2)

print(is_anagram('tops', 'stop'))
print(is_anagram('tops', 'stone'))
```

```python
word_list = open('words.txt').read().split()

for word in word_list:
    if is_anagram(word, 'takes'):
        print(word)
```

## 9.15.4. Exercise

- Write a `reverse_sentence` function that reverses the order of the words, adjusting capitalization. Example: `Reverse this sentence` → `Sentence this reverse`.

```python
def reverse_sentence(sentence):
    words = sentence.lower().split()
    return ' '.join(words[::-1]).capitalize()

print(reverse_sentence('Reverse this sentence'))
```

# Dictionaries

## 10.11.3. Exercise

- Write a `has_duplicates` function that takes a list or string and returns `True` when any element appears more than once. Can you think of a word longer than `unpredictably` without repeating a letter?

```python
def has_duplicates(sequence):
    seen = {}
    for item in sequence:
        if item in seen:
            return True
        seen[item] = 1
    return False

print(has_duplicates('unpredictably'))
print(has_duplicates('apple'))
print(has_duplicates([1, 2, 3, 2]))
```
