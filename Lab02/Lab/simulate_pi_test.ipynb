```python

import random
import math
import statistics

# Run one Monte Carlo simulation
def simulate_pi(convergence):
    n = 0  # Points inside the quarter-circle
    d = 0  # Total points
    pi = math.pi

    while True:
        # Generate random coordinates
        x = random.random()
        y = random.random()

        # Check whether point is inside the circle
        if x**2 + y**2 <= 1.0:
            n += 1

        # Update draw count and estimate of pi
        d += 1
        ratio = 4 * n / d

        # Calculate percentage difference
        percentage_difference = abs(ratio - pi) / pi * 100

        # Stop when convergence criterion is reached
        if percentage_difference <= convergence:
            return d


# Four convergence criteria from the assignment
criteria = [0.01, 0.001, 0.0001, 0.00001]

# Run 10 simulations for each criterion
for convergence in criteria:
    draws = []

    for run in range(10):
        draw_count = simulate_pi(convergence)
        draws.append(draw_count)

    # Calculate statistics
    average = statistics.mean(draws)
    standard_deviation = statistics.stdev(draws)

    # Display results
    print(f"Convergence criterion: {convergence}%")
    print(f"Draw counts: {draws}")
    print(f"Average draws: {average:.2f}")
    print(f"Standard deviation: {standard_deviation:.2f}")
    print("------------------------------")
```
