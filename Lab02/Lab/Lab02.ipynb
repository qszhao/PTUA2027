```python
def is_anagram(word1, word2):
    return sorted(word1) == sorted(word2)

print (is_anagram('tops', 'stops'))

word_list = ['tops','stops','takes','stake','steak','banana', 'apple']
             
for word in word_list:
    if is_anagram("takes", word) and word != "takes":
        print(word)
```

```python
def has_duplicates(sequence):
    seen = []
    for item in sequence:
        if item in seen:
            return True
        seen.append(item)
    return False

print(has_duplicates("subdermatoglyphic"))
```
