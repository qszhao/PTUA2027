```python
import math
import random

pi = math.pi

convergence = float(0.01)
sentinel = int(10000000)

n = 0
d = 0
simulating = True

while simulating:
    x = random.random()
    y = random.random()

    if x**2 + y**2 <= 1.0:
        n += 1
    d += 1

    ratio = 4.0 * n / d
    pct_diff = abs(ratio - pi) / pi * 100

    if pct_diff <= convergence:
        print(f"Converged at {d} draws!")
        print("Estimated Pi:", ratio)
        print("Percentage difference:", pct_diff, "%")
        break

    if d >= sentinel:
        print("Draw limit was exceeded!")
        print("Final estimated Pi:", ratio)
        print("Percentage difference from true Pi:", pct_diff, "%")
        break
```
