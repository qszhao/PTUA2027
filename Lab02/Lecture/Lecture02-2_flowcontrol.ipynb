## Python Conditions and If statements
Python supports the usual logical conditions from mathematics:

* Equals: a == b
* Not Equals: a != b
* Less than: a < b
* Less than or equal to: a <= b
* Greater than: a > b
* Greater than or equal to: a >= b
* These conditions can be used in several ways, most commonly in "if statements" and loops.


An "if statement" is written by using the if keyword.

```python
# if statement
a = 50
b = 25
if b > a:
    print("b is greater than a")
print("apple")
# Note: Indentation
# Python relies on indentation (whitespace at the beginning of a line) to define
# scope in the code. Other programming languages often use curly-brackets for this purpose.
```

```python
# Elif
# The elif keyword is pythons way of saying 
# "if the previous conditions were not true, then try this condition".
a = 50
b = 100
if b > a:
  print("b is greater than a")
elif a == b:
  print("a and b are equal")
```

```python
# Else
# The else keyword catches anything which isn't caught by the preceding conditions.
a = 100
b = 50
if b > a:
  print("b is greater than a")
elif a == b:
    print("a and b are equal")
    print("This is another line in the elif block")
else:
  print("a is greater than b")
```

```python
# You might have questions like:
# What's the difference between use if...elif...else vs multiple if statements?
# Answer: In if...elif...else structure, only one block will be executed. In multiple if statements, all conditions are checked and all true blocks are executed.
# Example:
score = 85

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B") 

# Here we only have one grade printed: Grade B. 
```

```python
# If we use multiple if statements, we will have two messages printed:
score = 95

if score >= 90:
    print("You get an A!")
if score >= 80:
    print("You passed the class!")
```

```python
# And
# The and keyword is a logical operator, and is used to combine conditional statements.
a = 200
b = 33
c = 100
if a > b and c > a:
  print("Both conditions are True")
else:
  print("we have a different situation")
```

```python
# Or
# The or keyword is a logical operator, and is used to combine conditional statements
a = 200
b = 33
c = 500
if a > b or a > c:
  print("At least one of the conditions is True")
```

```python
# Nested If
# You can have if statements inside if statements, this is called nested if statements.
x = 11

if x > 10:
    print("x is above ten")
    if x < 12:
        print("x is above ten and less than twelve")
else:
    print("x is less than ten")
```

## Python Loops
Python has two main loop commands:

* for loops
* while loops

## For Loop
A for loop is used for iterating over a sequence (that is either a list, a tuple, a dictionary, a set, or a string).

This is less like the for keyword in other programming languages, and works more like an iterator method as found in other object-orientated programming languages.

With the for loop we can execute a set of statements, once for each item in a list, tuple, or set, etc.

```python
# Print each student in a students list:
students = ["freshmen", "Sophomore", "Junior", "Senior"]
for xyz in students:
  print(xyz)
```

```python
# Looping Through a String
# Even strings are iterable objects, they contain a sequence of characters:
for x in "Senior":
  print(x)
```

```python
# The break Statement
# With the break statement we can stop the loop before it has looped through all the items.
# The break statement will only exit the innermost loop, see example below. 

# Exit the loop when x is "Sophomore"
students = ["freshmen", "Sophomore", "Junior", "Senior"]
for y in students:
  print(y)
  for x in students:
    if x == "Sophomore":
      break
    print(x)
# Question: why it only prints "freshmen" out?
```

```python
# The continue Statement
# With the continue statement we can stop the current iteration of the loop, and continue with the next.
# Do not print "Sophomore"
# Same as break, the continue statement will only affect the innermost loop.

students = ["freshmen", "Sophomore", "Junior", "Senior"]
for y in students:
  print(y)
  for x in students:
    if x == "Sophomore":
      continue
    print(x)
# Question: why it prints three words out?
```

```python
# The range() Function
# To loop through a set of code a specified number of times, we can use the range() function.
# The range() function returns a sequence of numbers, 
# starting from 0 by default, and increments by 1 (by default), 
# and ends at a specified number.
for x in range(5):
  print(x)
# Note that range(5) is not the values of 0 to 5, but the values 0 to 4.
```

```python
# you can also use a start number than the default 0
# and/or set a step in the increase of numbers (default is 1)
for x in range(2,20,2): # 2 is the start number
                        # 20 is the end number (this number-1)
                        # 2 is the increase step
      print(x)
```

```python
# Nested Loops
# A nested loop is a loop inside a loop.

# The "inner loop" will be executed one time 
# for each iteration of the "outer loop"

adj = ["a", "b", "c"]
fruits = [1,2,3,4]

for x in adj:
  print(x)
  for y in fruits:
    print(x, y)
```

## While loop

With the while loop we can execute a set of statements as long as a condition is true.

```python
# print i as long as i is less than 5:
i = 1
while i < 6:
  print(i)
  i += 1
# i += 1 means i = i+1
```

```python
# While loop support the same "break" and "continue" statement as for loop
i = 1
while i < 6: 
  if i == 2:
    break
  print(i)
  i += 1
```

```python
# While loop support the same "break" and "continue" statement as for loop
i = 0
while i < 6:
  
  if i == 3:
    continue
  i += 1
  print(i)
  
#Question: what's the difference between for loop and while loop when using continue? 
```

## Function
A function is a block of code which only runs when it is called.

You can pass data, known as parameters, into a function.

A function can return data as a result.

In Python a function is defined using the def keyword:


```python
# how to define a function.
# the function will not be executed before you call it
# function can have one or more parameters
def testFunction(a):
  if a > 3:
    print("Hello World")  
  else:
    print("Hello, ", a)
```

```python
# calling the function to run all statements within the function
testFunction(2)
```

```python
testFunction(4)
```

```python
# Default Parameter Value

# The following example shows how to use a default parameter value.
# If we call the function without parameter, it uses the default value:

def my_function(country = "UK"):
  print("I am from " + country)

my_function("China")
my_function("France")
my_function() # default value "UK" is used
my_function("USA")
```

```python
# Return Values
# To let a function return a value, use the return statement
def square(x=10):
  return x*x

print(square(3))
print(square(5))
print(square())
```
