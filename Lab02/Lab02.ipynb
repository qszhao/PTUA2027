```python
import math
# Part 1
# radius in centimeters (cm)
radius = 5
volume = ( 4 / 3) * math.pi * (radius ** 3)
print(volume)

# Part 2: cos^2(x) + sin^2(x) = 1 (x = 42)
x = 42
identity = math.cos(x)**2 + math.sin(x)**2
print(identity)

# Part 3: e^2
e_operator = math.e ** 2
e_power = math.pow(math.e, 2)
e_exp = math.exp(2)

print(f"Method 1 (math.e ** 2): {e_operator}")
print(f"Method 2 (math.pow)   : {e_power}")
print(f"Method 3 (math.exp)   : {e_exp}")
```

```python
def is_anagram(word1, word2):
    return sorted(word1) == sorted(word2)

word_list = []
for line in open('words.txt'):
    word = line.strip()
    word_list.append(word)

target = 'takes'
for word in word_list:
    if is_anagram(target, word):
        print(word)
```

```python
def reverse_sentence(sentence):
    words = sentence.split()
    reversed_words = reversed(words)
    new_sentence = ' '.join(reversed_words)
    return new_sentence.capitalize()
result = reverse_sentence('Reverse this sentence')
print(result)
```

```python
def has_duplicates(t):
    d = {}
    for x in t:
        if x in d: 
            return True
        d[x] = True
    return False
word_list = []
for line in open('words.txt'):
    word = line.strip()
    word_list.append(word)

for word in word_list:
    if len(word) > 13 and not has_duplicates(word):
        print(word, len(word))
```
